### Ejercicio 1 · Factura con descuento e IVA ★★

Pida el nombre de un producto, la cantidad y el precio unitario. Calcule:
- Subtotal = cantidad × precio
- Descuento del 10 % sobre el subtotal
- IVA del 15 % sobre (subtotal − descuento)
- Total

Muestre la factura con los valores **alineados a la derecha** en 10 espacios y 2 decimales:
```
Producto: Teclado
Subtotal:      90.00
Descuento:      9.00
IVA:           12.15
TOTAL:         93.15
```

````python
# Escriba su solución aquí

producto= input("ingrese el nombre del producto: ")
cantidad= int(input("ingrese la cantidad del producto: "))
precio_unidad= float(input("ingrese el precio por unidad: "))

subtotal= cantidad * precio_unidad
descuento= subtotal * 0.10 
descuento_con_subtotal= subtotal- descuento
iva= (subtotal - descuento)* 0.15
total= descuento_con_subtotal+iva

print(f"{'Producto:':<10} {producto}")
print(f"{'subtotal:':<10} {subtotal:>10.2f}")
print(f"{'descuento:':<10} {descuento:>10.2f}")
print(f"{'IVA:':<10} {iva:>10.2f}")
print(f"{'total:':<10} {total:>10.2f}")





````
